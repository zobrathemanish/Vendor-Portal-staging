```python
import pandas as pd

mapped = pd.read_parquet("mapped.parquet")
profiled = pd.read_parquet("data_profile.parquet")
canonical = pd.read_parquet("item_master_canonical.parquet")

print("Mapped rows:", len(mapped))
print("Profile rows:", len(profiled))
print("Canonical items:", canonical["_entity_id"].nunique())
```

```python
business_cols = [c for c in mapped.columns if not c.startswith("_")]

diff = mapped[business_cols].compare(
    profiled[business_cols],
    keep_shape=True,
    keep_equal=False
)

print("Profiling diffs:", diff.shape[0])
```
